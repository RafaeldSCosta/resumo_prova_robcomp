## 7. Estrutura mínima de um nó

Este código serve apenas para testar se o pacote foi configurado corretamente.

Posso colocar esta estrutura dentro de `q1.py` e depois completar com publishers, subscribers, controle do robô e o restante da questão.

```python
import rclpy
from rclpy.node import Node


class ProvaAI(Node):

    def __init__(self):
        # Cria o nó com o nome q1_node
        super().__init__('q1_node')

        # Mensagem para confirmar que o nó iniciou
        self.get_logger().info('Nó iniciado corretamente')


def main(args=None):
    # Inicia a comunicação com a ROS 2
    rclpy.init(args=args)

    # Cria o objeto do nó
    node = ProvaAI()

    # Mantém o nó funcionando
    rclpy.spin(node)

    # Destrói o nó quando o programa for encerrado
    node.destroy_node()

    # Encerra a ROS 2
    rclpy.shutdown()


if __name__ == '__main__':
    main()
```

## 15. Erros mais comuns

### `Package 'prova_ai' not found`

Provavelmente faltou compilar ou executar o `source`.

### `No executable found`

Verificar:

- se o executável foi adicionado ao `setup.py`;
- se o nome está correto;
- se o arquivo possui a função `main`;
- se compilei depois de alterar o `setup.py`;
- se executei o `source`.

### `ModuleNotFoundError`

Verificar:

- se escrevi corretamente o nome do módulo;
- se a dependência foi adicionada ao pacote;
- se os pacotes fornecidos pela disciplina estão atualizados;
- se compilei e executei o `source`.

### O simulador abriu, mas o robô não responde

Verificar:

- se meu nó está executando;
- se o nome do tópico está correto;
- se o tipo da mensagem está correto;
- se outro programa também está tentando controlar o robô;
- se meu código está realmente publicando velocidades.
